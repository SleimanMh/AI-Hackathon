```python
import kagglehub

path = kagglehub.dataset_download("pavansubhasht/ibm-hr-analytics-attrition-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv(path + "/WA_Fn-UseC_-HR-Employee-Attrition.csv")

print(df.info())
```

```python
columns_to_drop = ['EmployeeCount', 'EmployeeNumber', 'StockOptionLevel','Over18','StandardHours']
df = df.drop(columns=columns_to_drop)
print("Count of columns:", len(df.columns))
```

```python
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns.tolist()
print("Numerical columns:", numerical_features)
count = len(numerical_features)
print("Count of numerical columns:", count)
```

```python
categorical_features = df.select_dtypes(include=['object']).columns.tolist()
print("Categorical columns:", categorical_features)
count = len(categorical_features)
print("Count of categorical columns:", count)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

for col in categorical_features:
    print(f"{col}: {df[col].unique()}\n")
```

```python
from sklearn.preprocessing import LabelEncoder

encoders = {}
for col in categorical_features:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    encoders[col] = le  

for col in categorical_features:
    print(f"{col}: {list(df[col].unique())}")
```

```python
from sklearn.preprocessing import MinMaxScaler
min_max_scaler = MinMaxScaler()

original_columns = df.columns
df = min_max_scaler.fit_transform(df)

df = pd.DataFrame(df, columns=original_columns)
df.head()
```

```python
import numpy as np
# Skewness for features
skewness = df.select_dtypes(include=[np.number]).skew()
print("Skewness for each feature:\n", skewness)
```

```python
high_skew = skewness[abs(skewness) > 1].index
print("Highly skewed features:", list(high_skew))

for col in high_skew:
    # Add 1 to avoid log(0)
    df[col] = np.log1p(df[col])
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Select continuous features
continuous_features = ['MonthlyIncome', 'JobSatisfaction', 'Age', 'YearsAtCompany']

# Plot distributions
for feature in continuous_features:
    plt.figure(figsize=(8, 5))
    sns.histplot(df[feature].dropna(), kde=True, bins=30)
    plt.title(f'Distribution of {feature}')
    plt.xlabel(feature)
    plt.ylabel('Frequency')
    plt.show()
```

```python
sns.pairplot(df[['MonthlyIncome', 'Age','Attrition','JobSatisfaction','TotalWorkingYears','YearsInCurrentRole']].dropna(),
                diag_kind='kde', hue='Attrition')
plt.suptitle('Feature Interactions', y=1.02)
plt.show()
        
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the correlation matrix
correlation_matrix = df.corr()

# Create a heatmap to visualize the correlation
plt.figure(figsize=(10, 8))  # Adjust the figure size for better readability
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)

# Add titles and labels
plt.title('Correlation Heatmap for Numerical Features', fontsize=16)
plt.tight_layout()
plt.show()
```

```python
sns.violinplot(data=df, x='Attrition', y='MonthlyIncome', palette='coolwarm', split=True)
plt.title('Monthly Income Distribution by Attrition')
plt.xlabel('Attrition (0 = No, 1 = Yes)')
plt.ylabel('Monthly Income')
plt.show()
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
import matplotlib.pyplot as plt

# Step 1: Prepare the target and features
# Use double brackets to ensure X is a DataFrame, not a Series
feature_name = 'MonthlyIncome'  # Change this to the feature you want to analyze
X = df[[feature_name]].dropna()
y = df.loc[X.index, 'Attrition']  # Align y with X after dropna

# Step 2: Apply LDA
lda = LinearDiscriminantAnalysis(n_components=1)  # Reduce to a single component for binary target
# Convert y to binary (0 or 1) for LDA
y_binary = y.apply(lambda v: 1 if v > 0 else 0)
attrition_lda = lda.fit_transform(X, y_binary)

# Step 3: Visualize the LDA result
plt.scatter(attrition_lda, y, c=y, cmap='coolwarm')
plt.title(f'LDA of {feature_name} vs Attrition')
plt.xlabel('Linear Discriminant 1')
plt.ylabel('Attrition')
plt.colorbar(label='Attrition')
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, average_precision_score, classification_report, roc_auc_score
from xgboost import XGBClassifier

X = df[[col for col in categorical_features + numerical_features if col != "Attrition"]]
y = df["Attrition"].apply(lambda v: 1 if v > 0 else 0)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)

clf = XGBClassifier(
    n_estimators=500,
    max_depth=6,
    learning_rate=0.1,
    subsample=0.9,
    colsample_bytree=0.9,
    gamma=0.1,
    min_child_weight=1,
    reg_alpha=0.01,
    reg_lambda=1,
    scale_pos_weight=1,
    random_state=42,
    use_label_encoder=False,
    eval_metric='logloss'
)

clf.fit(X_train, y_train)

# Print the exact order of columns used for fitting
print('ORDER FOR app.py:')
print(list(X.columns))

y_pred = clf.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```
